Defining a Class for Stock Objects

```python
class Stock:
    def __init__(self, exchange, symbol):
        self.exchange = exchange
        self.symbol = symbol
```

Create a list of Stock Objects

```python
stocks = []
```

```python
# OK
def stock_info(stock_str):
  step = 1
  exchange = ""
  symbol = ""
  for char in stock_str:
    if step == 1:
      if char != ':':
        exchange += char
      else:
        step = 2
    else:
      if char != ',':
        symbol += char
      else:
        step = 1
  return exchange, symbol

# OK
# Store stock objects
stock_objects = []
stock_str = ""
for char in list_downloaded:
  if char != ',':
    stock_str = stock_str + char
  else:
    exchange, symbol = stock_info(stock_str)
    stock_objects.append(Stock(exchange, symbol))
    stock_str = ""

# OK
# for stock in stock_objects:
#   print(f"{stock.symbol}: {stock.exchange}")
```
