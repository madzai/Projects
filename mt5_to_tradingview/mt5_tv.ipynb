Get MT5 US stock symbols from csv file

```python
import pandas as pd

# Reading the CSV file
data = pd.read_csv("mt5_symbols.csv")

# Converting specific columns into lists
symbols = data['Symbol'].tolist()
```

```python
mt5_us = ""

for symbol in symbols:
  mt5_us += symbol + ','

# remove last char (comma)
mt5_us = mt5_us[:-1]
print(mt5_us)
```
